```python
## Original Aufgabe

def manipulate_list(lst: list[int]) -> list[int]:
    """Gibt die geraden Zahlen aus lst zurück, jeweils verdoppelt."""
    list = []
    for number in lst: 
        if number % 2 == 0:
            list.append(number * 2)
        
    return list
```

```python
def filter_scores(scores):
    liste = []

    for score in scores:
        if score > 10:
            liste.append(score - 5)
    return liste
filter_scores([5, 12, 18, 10, 25])
```

```python
def process_numbers(numbers):
    liste = []
    for num in numbers:
        if num < 20:
            liste.append(num * 3)
    return liste 
process_numbers([5, 10, 15, 20, 25])
```

```python
def filter_words(words):
    liste = []
    for word in words:
        if len(word) >= 5:
            liste.append(word)
    return liste
filter_words(["Apfel", "Hi", "Banane", "Haus", "Computer"])
```

```python
def count_positive(numbers):
    liste = []
    for num in numbers:
        if num >=0:
            liste.append(num)
    return liste
count_positive([-3, 0, 5, 8, -1, 2])
```

```python
def reverse_list(numbers):
    liste = []
    for i in numbers:
        liste.append(i)
    liste.reverse()
    return liste

reverse_list([1,2,3,4,5])
        
```

```python
def double_numbers(numbers):
    liste = []
    for i in numbers:
        liste.append(i * 2)
    return liste
double_numbers([18, 19, 23, 4])
```

```python
def person_info(person):
    p1 = person[0]
    p2 = person[1]
    p3 = person[2]

    return f'Person {p1} ist {p2} Jahre alt und ist {p3} m gross'

print(person_info(("James", 12, 1.23)))
```

```python
def tuple_sum(tpl):
    sum = 0
    for element in tpl:
        sum+=element
    return sum
tuple_sum((10,10,10))
```

```python
def check_element(tpl, item):
    for i in tpl:
        if i == item:
            return True
    return False
check_element(('a', 'b', 'c'), 'd')
    
```

```python
def reverse_tuple(tpl):
    return tpl[::-1]
    
    
reverse_tuple((1,2,3,4,5))
```

```python
def first_and_last(tpl):
    return (tpl[0], tpl[-1])

print(first_and_last(("Start", "Mitte1", "Mitte2", "Ende")))
# Output: ('Start', 'Ende')
```

```python
def tuple_info(tpl):
    count = len(tpl)
    return f'Der tuple enthält {count} Elemente und das erste ist vom Typ ({type(tpl[0])})'
tuple_info((42, "Hallo", 3.5))
```

```python
def modify_tuple(tpl, index, new_value):
    lst = list(tpl)

    lst[index] = new_value

    return tuple(lst)
modify_tuple((1, 2, 3), 1, 99)
```

```python
def count_item(tpl, item):
    count = 0
    for num in tpl:
        if num == item:
            count = count+1
    return count
count_item((1, 2, 3, 2, 4, 2), 2)
```

```python
def get_inner_value(tpl):
    return tpl[0][1]
get_inner_value(((1, 5), (3, 8)))
```

```python
def merge_tuples(tpl1, tpl2):
    return tpl1 + tpl2
merge_tuples((1, 2), (3, 4))
```

```python
def contains_element(tpl, item):
    for i in tpl:
        if i == item:
            return True
    return False
contains_element((10, 20, 30), 210)
```

```python
def get_intersection(tpl1, tpl2):
    return set(tpl1) & set(tpl2)
get_intersection((1, 2, 3, 4), (3, 4, 5, 6))
```

```python
def get_passed(scores):
    return {key : value for key, value in scores.items() if value >=50}
get_passed({"Anna": 80, "Ben": 35, "Clara": 50, "David": 42})
```

```python
def calculate_weighted_scores(names, scores, factor):
    ergebnis = dict(map(lambda x: (x[0],x[1] * factor, zip(names, scores)))
    return ergebnis

    
```

```python
class Product():
    def __init__(self, name, price):
        self.name = name 
        self.price = price
    def discount(self, percent):
        self.price = self.price * (1 - percent / 100)
        return self.price

    def __eq__(self, other):
        if not isinstance ( other, Product):
            return False
        return self.name == other.name and  self.price == other.price
    def __str__(self): 
        return f'{self.name}:{self.price}'
p1 = Product("Tastatur", 100)
p2 = Product("Tastatur", 100)

p1.discount(10)  # Preis von p1 wird 90
p1 == p2         # False
str(p1)          # "Tastatur: 90"
```
